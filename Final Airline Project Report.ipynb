**Departure Period**

```python
def get_time_period(hour):
  if hour<6:
    return "early morning"
  elif hour<12:
    return "morning"
  elif hour<18:
    return "afternoon"
  else:
    return "night"

df["departurePeriod"]=df["departurehour"].apply(get_time_period)
df["departurePeriod"].value_counts()
```

Airline Performance Calculation

```python
# Flight schedule count from our main dataset
schedule_metrics = (
    df.groupby("airline")
    .size()
    .reset_index(name="scheduled_flights")
)

# Government performance metrics
gov_metrics = gov_df[
    ["airline", "cancellation_rate_percent", "avg_otp_jan_may"]
].copy()

# Standardize airline names
gov_metrics["airline"] = gov_metrics["airline"].replace({
    "Spice Jet": "SpiceJet",
    "Alliance Air": "Alliance Air (India)",
    "Indiaone Air": "IndiaOne Air"
})

# Merge both datasets
airline_performance = schedule_metrics.merge(
    gov_metrics,
    on="airline",
    how="inner"
)

# Sort by OTP
airline_performance = airline_performance.sort_values(
    "avg_otp_jan_may",
    ascending=False
)

display(airline_performance)
```

Business Insight: Akasa Air shows the strongest performance with the highest OTP (80.86%) and lowest cancellation rate (0.14%). IndiGo has the largest schedule volume (18,991 flights), while Alliance Air shows relatively lower OTP (51.22%) and higher cancellations (3.82%).

OTP vs Cancellation Rate Visualization.

```python
import matplotlib.pyplot as plt

plot_df = airline_performance.dropna(
    subset=["avg_otp_jan_may"]
).copy()

plt.figure(figsize=(10, 6))

plt.scatter(
    plot_df["cancellation_rate_percent"],
    plot_df["avg_otp_jan_may"]
)

for _, row in plot_df.iterrows():
    plt.annotate(
        row["airline"],
        (row["cancellation_rate_percent"],
         row["avg_otp_jan_may"]),
        xytext=(5, 5),
        textcoords="offset points"
    )

plt.xlabel("Cancellation Rate (%)")
plt.ylabel("Average OTP (%)")
plt.title("Airline OTP vs Cancellation Rate")
plt.grid(True, alpha=0.3)

plt.show()
```

Business Insight: Akasa Air has the best operational performance with high OTP and the lowest cancellation rate. Alliance Air shows the weakest performance, with lower OTP and a higher cancellation rate.

Scheduled Flights vs Average OTP

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    plot_df["scheduled_flights"],
    plot_df["avg_otp_jan_may"]
)

for _, row in plot_df.iterrows():
    plt.annotate(
        row["airline"],
        (row["scheduled_flights"], row["avg_otp_jan_may"]),
        xytext=(5, 5),
        textcoords="offset points"
    )

plt.xlabel("Scheduled Flights")
plt.ylabel("Average OTP (%)")
plt.title("Scheduled Flights vs Average OTP")
plt.grid(True, alpha=0.3)

plt.show()
```

Business Insight: IndiGo handles the highest flight volume while maintaining a relatively strong OTP of 72.64%. Akasa Air and Vistara achieve higher OTP with much lower schedules, indicating that maintaining punctuality can become more challenging as flight volume increases.

Final Airline Performance Ranking

```python
ranking_df = airline_performance.dropna(
    subset=["avg_otp_jan_may"]
).copy()

# Rank airlines by OTP
ranking_df["otp_rank"] = (
    ranking_df["avg_otp_jan_may"]
    .rank(ascending=False, method="dense")
    .astype(int)
)

ranking_df = ranking_df.sort_values("otp_rank")

display(
    ranking_df[
        ["otp_rank",
          "airline",
          "scheduled_flights",
          "avg_otp_jan_may",
          "cancellation_rate_percent"
        ]
    ]
)
```

Akasa Air ranks highest with 80.86% average OTP and the lowest cancellation rate of 0.14%, indicating strong operational reliability. Government + Flight Dataset combine

```python
final_df = airline_metrics.merge(
    gov_df[
        [
            'airline_standard',
            'total_scheduled_departures_2024_till_may',
            'flights_cancelled_2024_till_may',
            'cancellation_rate_percent',
            'avg_otp_jan_may'
        ]
    ],
    left_on='airline',
    right_on='airline_standard',
    how='inner'
)

# Select useful columns
final_df = final_df[
    [
        'airline',
        'total_schedules',
        'average_flight_duration',
        'avg_operating_days',
        'avg_departure_hour',
        'total_scheduled_departures_2024_till_may',
        'flights_cancelled_2024_till_may',
        'cancellation_rate_percent',
        'avg_otp_jan_may'
    ]
]

final_df
```

The combined analysis shows that IndiGo has the largest operational scale, while Akasa Air delivers the strongest reliability with high OTP and the lowest cancellation rate.

Higher operational scale does not necessarily mean better reliability, as Akasa Air outperforms larger airlines on OTP and cancellation rate.

Overall Airline Performance Score

```python
performance_df = final_df.dropna(subset=['avg_otp_jan_may']).copy()

# Performance score
performance_df['performance_score'] = (
    performance_df['avg_otp_jan_may']
    - (performance_df['cancellation_rate_percent'] * 5)
)

# Rank airlines
performance_df['performance_rank'] = (
    performance_df['performance_score']
    .rank(ascending=False, method='dense')
    .astype(int)
)

# Final ranking
performance_df = performance_df.sort_values('performance_rank')

performance_df[
    [
        'performance_rank',
        'airline',
        'avg_otp_jan_may',
        'cancellation_rate_percent',
        'performance_score'
    ]
]
```

Akasa Air ranks highest on the custom performance score, combining the highest OTP with the lowest cancellation rate, while Alliance Air ranks lowest among airlines with available OTP data.

The score is an analytical metric created from OTP and cancellation rate, not an official government ranking.

Scale vs Reliability

```python
scale_reliability = final_df.dropna(
    subset=['avg_otp_jan_may']
).copy()

scale_reliability[
    [
        'airline',
        'total_scheduled_departures_2024_till_may',
        'avg_otp_jan_may',
        'cancellation_rate_percent'
    ]
].sort_values(
    'total_scheduled_departures_2024_till_may',
    ascending=False
)
```

Operational scale does not directly translate into better reliability; Akasa Air has much lower flight volume than IndiGo but achieves higher OTP and a significantly lower cancellation rate.

Final Top/Bottom Performance

```python
final_ranking = performance_df[
    [
        'performance_rank',
        'airline',
        'total_scheduled_departures_2024_till_may',
        'avg_otp_jan_may',
        'cancellation_rate_percent',
        'performance_score'
    ]
].sort_values('performance_rank')

final_ranking
```
